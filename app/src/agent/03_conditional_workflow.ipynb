```python
from typing import TypedDict, Literal
from dotenv import load_dotenv
from langchain_core.messages import SystemMessage, HumanMessage
from langchain_openai import ChatOpenAI
from langgraph.graph import StateGraph, START, END
load_dotenv()  # Load environment variables from .env file
```

```python
class State(TypedDict):
    message: str
    category: str
    reply: str
```

```python
def classify(state: State):
    text = state["message"].lower()
    if any(w in text for w in ("invoice", "refund", "charge")):
        category = "billing"
    elif any(w in text for w in ("error", "crash", "bug")):
        category = "technical"
    else:
        category = "general"
    return {"category": category}

# 2. The router: reads state, returns a label, changes nothing
def route(state: State) -> Literal["billing", "technical", "general"]:
    return state["category"] # type: ignore

# 3. The destination nodes
def billing(state: State):   return {"reply": "Routing you to billing. Refund policy: ..."}
def technical(state: State): return {"reply": "Sorry about the error. Please share your logs."}
def general(state: State):   return {"reply": "Thanks for reaching out! How can we help?"}
```

```python
g = StateGraph(State)
g.add_node("classify", classify)
g.add_node("billing", billing)
g.add_node("technical", technical)
g.add_node("general", general)

g.add_edge(START, "classify")
g.add_conditional_edges(
    "classify",            # run the router after this node
    route,                 # function that picks the path
    {"billing": "billing", "technical": "technical", "general": "general"},
)
for node in ("billing", "technical", "general"):
    g.add_edge(node, END)

g.compile()
```
